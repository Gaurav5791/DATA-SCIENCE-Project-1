```python
#Without Legends 2 line graph

import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr"]

sales_2024 = [100,150,120,180]

sales_2025 = [120,160,140,200]

plt.plot(months,sales_2024)

plt.plot(months,sales_2025)

plt.show()
```

```python
#Using Legends 2 line graph

import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr"]

sales_2024 = [100,150,120,180]

sales_2025 = [120,160,140,200]

plt.plot(months,sales_2024,label="Sales 2024")

plt.plot(months,sales_2025,label="Sales 2025")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))

months = ["Jan","Feb","Mar","Apr"]

sales_2024 = [100,150,120,180]

sales_2025 = [120,160,140,200]

plt.plot(months,sales_2024,
         label="Sales 2024",
         marker="o")

plt.plot(months,sales_2025,
         label="Sales 2025",
         marker="*")

plt.title("Sales Comparison")

plt.xlabel("Months")

plt.ylabel("Sales")

plt.grid()

plt.legend()

plt.show()
```
