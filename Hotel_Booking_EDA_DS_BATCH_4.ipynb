# Phase 3: Data Quality Assessment

## What is Data Quality Assessment?

Data quality assessment is a **systematic evaluation** of your dataset for errors, inconsistencies, incompleteness, and invalid values. It answers the question: **"Can I trust this data?"**

Real-world data is almost never clean. It comes from multiple systems, entered by humans, and subject to technical errors. Before drawing conclusions, you must understand the reliability of your raw material.

## The Five Dimensions of Data Quality

| Dimension | Definition | Example Issue |
|-----------|-----------|--------------|
| **Completeness** | Are all values present? | Missing country for many bookings |
| **Validity** | Are values within expected ranges? | ADR = -10 (impossible) |
| **Consistency** | Are values logically consistent? | is_canceled=0 but status="Canceled" |
| **Uniqueness** | Are there duplicate records? | Same booking appears twice |
| **Accuracy** | Do values reflect reality? | 26 adults in one room |

## Why Does Data Quality Matter?

Poor data quality produces poor insights — this is the principle of **"Garbage In, Garbage Out" (GIGO)**.

Consider: if 15% of ADR values are zero due to a system error, and you compute mean ADR without investigating, your revenue figures will be significantly underestimated. Decisions based on those figures could lead to pricing strategies that cost the business real money.

| `df[col].value_counts()` | Frequency of each unique value |
| `df[df[col] < 0]` | Filter rows where a value is invalid |

# Phase 4: Data Cleaning

## What is Data Cleaning?

Data cleaning is the process of **detecting and correcting errors, inconsistencies, and invalid values** in a dataset. It transforms raw, messy data into a reliable foundation for analysis.

## Why is Data Cleaning Critical?

Every decision made using this data depends on its quality. A single uncorrected error — like leaving in duplicate bookings — can cascade through the entire analysis, producing metrics that look plausible but are factually wrong.

> *"Data scientists spend 60–80% of their time cleaning data — and yet it is often the most overlooked skill in data science education."*

## Cleaning vs Removing: When to Use Each

| Scenario | Recommended Approach |
|----------|---------------------|
| Missing value in a low-importance column, <1% missing | Fill with mode, median, or 'Unknown' |
| Missing value in a critical column, >50% missing | Consider dropping the column |
| Small number of duplicate rows | Remove exact duplicates |
| Invalid values (negative price) | Remove or cap |
| Wrong data type (date as string) | Convert |
| Inconsistent category labels | Standardise |

## The Cleaning Workflow

Always apply cleaning in this order:

1. **Work on a copy** — `df_clean = df.copy()` — never modify the original
2. **Handle duplicates first** — removing duplicates changes row counts used in subsequent steps
3. **Handle missing values** — per-column, with documented justification
4. **Fix data types** — convert to correct types after handling missings
5. **Fix invalid values** — remove or correct logically impossible values
6. **Standardise categories** — consistent capitalisation, unified labels
7. **Document every decision** — record what you changed and why

## Key Functions for Data Cleaning

| Function | Purpose |
|----------|---------|
| `df.copy()` | Create an independent copy of the DataFrame |
| `df.drop_duplicates()` | Remove exact duplicate rows |
| `df[col].fillna(value)` | Fill missing values with a specified value |
| `df.dropna(subset=[col])` | Drop rows with missing values in specific columns |
| `df[col].astype(dtype)` | Convert a column to a different data type |
| `df[col].str.strip()` | Remove leading/trailing whitespace |
| `df[col].str.title()` | Standardise capitalisation |
| `df.rename(columns={old: new})` | Rename columns |
| `df[col].replace({old: new})` | Replace specific values |

# Phase 5: Univariate Analysis

## What is Univariate Analysis?

Univariate analysis examines **one variable at a time**. It is the most fundamental form of data exploration — before you can understand how variables relate to each other, you must understand each one individually.

*Uni* = one. *Variate* = variable.

## Why Do We Perform Univariate Analysis?

Univariate analysis answers questions like:
- What is the typical (central) value?
- How spread out are the values?
- Are values concentrated in a few categories, or evenly distributed?
- Are there extreme values that dominate the distribution?
- What shape does the distribution take?

## Measures of Central Tendency

| Measure | What It Tells You | When to Use |
|---------|------------------|------------|
| **Mean** | Arithmetic average | When data is roughly symmetric and has no extreme outliers |
| **Median** | Middle value when sorted | When data is skewed or has outliers (more robust than mean) |
| **Mode** | Most frequently occurring value | For categorical data or discrete numerical data |

## Measures of Spread (Dispersion)

| Measure | What It Tells You |
|---------|------------------|
| **Range** | Max − Min: total span of values |
| **Variance** | Average squared deviation from the mean |
| **Standard Deviation** | Average deviation from the mean (same units as data) |
| **IQR (Interquartile Range)** | Q3 − Q1: spread of the middle 50% |

## Visualisations for Numerical Variables

| Chart | Purpose | When to Use |
|-------|---------|------------|
| **Histogram** | Shows frequency of value ranges (bins) | Always use for first look at a numerical distribution |
| **KDE Plot** | Smoothed version of a histogram | When you want a continuous distribution curve |
| **Boxplot** | Shows quartiles, median, and outliers in one chart | When you want to see spread AND outliers simultaneously |

## Visualisations for Categorical Variables

| Chart | Purpose | When to Use |
|-------|---------|------------|
| **Count Plot** | Frequency of each category as bars | For categorical variables with few unique values |
| **Bar Plot** | Display of aggregated values per category | When comparing a metric across categories |
| **Pie Chart** | Proportional share of each category | Use sparingly — only for 2–4 categories |

## Reading a Boxplot

```
|----[  Q1 | MEDIAN | Q3  ]----| ← Whiskers → ●  ●  (Outliers)
```
- **Box:** Middle 50% of the data (IQR)
- **Line inside box:** Median
- **Whiskers:** Q1 − 1.5×IQR to Q3 + 1.5×IQR
- **Dots beyond whiskers:** Outliers

## Functions Cheat Sheet

| Function | Usage |
|----------|-------|
| `df[col].describe()` | Count, mean, std, min, quartiles, max |
| `df[col].mean()` | Mean |
| `df[col].median()` | Median |
| `df[col].mode()` | Mode |
| `df[col].std()` | Standard deviation |
| `df[col].var()` | Variance |
| `df[col].quantile([.25, .5, .75])` | Specific percentiles |
| `df[col].value_counts()` | Frequency table for categorical |
| `df[col].value_counts(normalize=True)` | Proportions |
| `sns.histplot(df[col], kde=True)` | Histogram with KDE overlay |
| `sns.boxplot(y=df[col])` | Boxplot |
| `sns.countplot(data=df, x=col)` | Countplot for categorical |

# Phase 6: Outlier Detection

## What is an Outlier?

An outlier is a data point that **differs significantly from the majority** of observations. The term comes from statistics, where values that fall far from the "centre" of a distribution are considered to lie *outside* the normal range.

## Why Does Outlier Detection Matter?

Outliers can affect your analysis in several ways:
- **Statistical distortion:** A single ADR of €5,000 can pull the mean far above the typical booking value
- **Model skewing:** Predictive models often perform poorly when trained on data with extreme outliers
- **Signal vs. noise:** Sometimes outliers are the most interesting data points (VIP customers, fraudulent bookings)
- **Decision making:** Managers making pricing decisions based on outlier-inflated averages will missprice rooms

## The Most Important Question

> **Is this outlier an error, or is it a real extreme value?**

This question has no universal answer. Context matters enormously:
- An ADR of €2,000 might be an error — or it might be a genuine penthouse suite booking
- A lead time of 737 days might be a system error — or it might be a wedding venue booked 2 years in advance
- 26 previous cancellations might be a data corruption — or it might be a chronic problem guest

## Three Methods for Detecting Outliers

### Method 1: Boxplot (Visual Method)
The boxplot is the most immediate visual tool. Any point beyond the whiskers is flagged as a potential outlier.
- **Whisker upper bound:** Q3 + 1.5 × IQR
- **Whisker lower bound:** Q1 − 1.5 × IQR

### Method 2: IQR Method (Statistical Method)
The Interquartile Range method is the numerical equivalent of the boxplot:
```
IQR = Q3 - Q1
Lower Fence = Q1 - 1.5 × IQR
Upper Fence = Q3 + 1.5 × IQR
Any value outside [Lower Fence, Upper Fence] is a potential outlier
```

### Method 3: Z-Score Method
Z-score measures how many standard deviations a value is from the mean:
```
Z = (value - mean) / standard_deviation
|Z| > 3 is commonly used as the outlier threshold
```
- **Advantage:** Intuitive interpretation ("this value is 4 standard deviations above average")
- **Disadvantage:** Assumes roughly normal distribution; sensitive to extreme values pulling the mean

## When to Remove Outliers

| Situation | Action |
|-----------|--------|
| Clearly a data entry error (negative age, impossible value) | Remove |
| Very small number of outliers with no business explanation | Remove or cap |
| Outlier represents a legitimate extreme case (VIP, group booking) | Keep, but analyse separately |
| Outlier affects only one analysis (e.g., mean ADR) | Use median instead of mean |

## Functions Used

| Function | Purpose |
|----------|---------|
| `sns.boxplot(y=df[col])` | Visual outlier detection |
| `df[col].quantile(0.25)` | Q1 |
| `df[col].quantile(0.75)` | Q3 |
| `scipy.stats.zscore(df[col])` | Z-score for each value |
| `np.abs(z_scores) > 3` | Identify outliers by Z-score |

# Phase 7: Bivariate Analysis

## What is Bivariate Analysis?

Bivariate analysis examines the **relationship between two variables**. Where univariate analysis describes individual variables, bivariate analysis reveals how they *interact*.

*Bi* = two. *Variate* = variable.

## Why is Bivariate Analysis Important?

Most business questions involve relationships:
- "Do guests who book further in advance cancel more?" → `lead_time` vs `is_canceled`
- "Which room types command the highest price?" → `reserved_room_type` vs `adr`
- "Are guests from certain countries more likely to cancel?" → `country` vs `is_canceled`

Answering these questions requires comparing two variables simultaneously.

## The Three Types of Bivariate Analysis

### Type 1: Numerical vs. Numerical

Used when both variables are numbers (e.g., `lead_time` vs `adr`).

| Tool | Purpose |
|------|---------|
| `df.corr()` | Pearson correlation coefficient: −1 to +1 |
| `sns.scatterplot()` | Visual relationship between two numbers |
| `sns.regplot()` | Scatter with regression line |
| `sns.jointplot()` | Combined scatter + distribution |

**Reading Correlation:**
- `r ≈ +1`: Strong positive relationship (as X increases, Y increases)
- `r ≈ −1`: Strong negative relationship (as X increases, Y decreases)
- `r ≈ 0`: No linear relationship

⚠️ **Correlation ≠ Causation.** Two variables can be correlated without one causing the other.

### Type 2: Categorical vs. Numerical

Used to compare a metric across groups (e.g., `hotel` vs `adr`).

| Tool | Purpose |
|------|---------|
| `df.groupby(cat)[num].mean()` | Mean of numerical variable per category |
| `sns.boxplot(x=cat, y=num)` | Distribution per group |
| `sns.barplot(x=cat, y=num)` | Mean per group with confidence interval |

### Type 3: Categorical vs. Categorical

Used to compare the frequency of one category against another (e.g., `market_segment` vs `is_canceled`).

| Tool | Purpose |
|------|---------|
| `pd.crosstab(cat1, cat2)` | Frequency table of all combinations |
| `pd.crosstab(cat1, cat2, normalize='index')` | Row percentages |
| `sns.countplot(x=cat1, hue=cat2)` | Grouped bar chart |
| `sns.heatmap(crosstab)` | Colour-coded crosstab |

# Phase 8: Multivariate Analysis

## What is Multivariate Analysis?

Multivariate analysis examines **three or more variables simultaneously**. It reveals patterns and interactions that are invisible when you look at variables in isolation or in pairs.

*Multi* = many. *Variate* = variable.

## Why Go Beyond Bivariate Analysis?

Consider this scenario: You find that "Online TA" bookings have a high cancellation rate, AND that "No Deposit" bookings have a high cancellation rate. But are they the *same* bookings? Multivariate analysis lets you ask: among Online TA bookings with no deposit, what is the cancellation rate? The answer might be 70% — far more actionable than either finding alone.

Reality is multivariate. Business decisions are made in multivariate contexts.

## Techniques in This Phase

### 1. Correlation Matrix + Heatmap
A correlation matrix shows the pairwise correlation between *all* numerical variables simultaneously. Visualised as a heatmap, it immediately reveals which pairs are strongly related.

```python
df.corr()                    # Compute correlation matrix
sns.heatmap(df.corr(), ...)  # Visualise as heatmap
```

**Interpretation:**
- Deep red = strong positive correlation
- Deep blue = strong negative correlation
- White/light = no meaningful correlation

### 2. Pair Plot
A pair plot creates a grid of scatter plots for all combinations of selected variables. The diagonal typically shows the distribution of each variable individually.

```python
sns.pairplot(df[selected_cols], hue='category_col')
```

**Use when:** You want a comprehensive overview of all pairwise relationships in one chart.

### 3. Pivot Tables
A pivot table aggregates data across two categorical dimensions simultaneously.

```python
pd.pivot_table(df, values='adr', index='hotel',
               columns='market_segment', aggfunc='mean')
```

**Use when:** You want to compare a metric across two categorical variables at once (e.g., ADR by hotel type AND market segment).

### 4. Multi-Level Crosstabs
Cross-tabulations can reveal interaction effects between three variables:

```python
pd.crosstab(df['market_segment'], df['deposit_type'],
            values=df['is_canceled'], aggfunc='mean')
```

### 5. Grouped Heatmaps
Heatmaps can show cancellation rates or ADR across two dimensions, immediately revealing which combinations are problematic.

# Phase 9: Feature Engineering

## What is Feature Engineering?

Feature engineering is the process of **creating new variables from existing ones** to improve the analytical power of your dataset. It transforms raw data into representations that better capture the underlying business logic.

## Why Do Analysts Engineer Features?

Raw data columns often don't directly represent the concepts you care about. Consider:
- `stays_in_week_nights` and `stays_in_weekend_nights` are useful individually — but `total_nights` is simpler and more intuitive for most analyses
- `arrival_date_month` is a specific date attribute — but `arrival_season` is a higher-level business concept that is more useful for trend analysis
- `reserved_room_type != assigned_room_type` is a logical expression — but converting it to a `room_change_flag` binary variable makes it immediately usable in analysis

Feature engineering is both a **technical** and **domain knowledge** skill. The best features come from understanding the business deeply.

## Types of Engineered Features

| Type | Description | Example |
|------|-------------|---------|
| **Aggregation** | Summing or averaging related columns | `total_guests = adults + children + babies` |
| **Extraction** | Pulling a component from a complex column | `arrival_quarter` from `arrival_date_month` |
| **Binary Flag** | Converting a condition to 0/1 | `room_change_flag = (reserved ≠ assigned)` |
| **Ratio / Rate** | Dividing two related quantities | `weeknight_ratio = week_nights / total_nights` |
| **Categorisation** | Grouping continuous values into bins | `lead_time_bucket: Short/Medium/Long` |

## Business Value of Each Feature

The features you create in this phase will be used directly in Phase 10 visualisations and Phase 11 findings. Each one should represent a **meaningful business concept**, not just a mathematical manipulation.
