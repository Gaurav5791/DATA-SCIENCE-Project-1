```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
```

```python
df1=df.drop_duplicates(inplace=True)
df1
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['age']
```

```python
df.info()
```

```python
df
```

```python
df.drop(columns=['deck'], inplace=True)
```

```python
df
```

```python
df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'].fillna(df['embark_town'].mode()[0])
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df['survived'] = df['survived'].astype(int)
```

```python
df['survived'].sum()
```

```python
print(df.isnull().sum())
```

```python
print("\nShape after cleaning:", df.shape)
```

```python
survival_rate = np.mean(df['survived']) * 100
print(survival_rate,"%")
```

```python
avg_age_survived = np.mean(df[df['survived'] == 1]['age'])
print(avg_age_survived)
```

```python
avg_age_not_survived = np.mean(df[df['survived'] == 0]['age'])
print(avg_age_not_survived)
```

```python
print("\nFare Percentiles:")
print(f"25th percentile: {np.percentile(df['fare'], 25):.2f}")
print(f"50th percentile: {np.percentile(df['fare'], 50):.2f}")
print(f"75th percentile: {np.percentile(df['fare'], 75):.2f}")
print(f"90th percentile: {np.percentile(df['fare'], 90):.2f}")
```

```python
print("Survival Rate by Gender:\n",df.groupby('sex')['survived'].mean() * 100)
```

```python
print("Survival Rate by passenger class :\n",df.groupby('pclass')['survived'].mean() * 100)
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6, 4))
survival_counts = df['survived'].value_counts()
plt.bar(['Not Survived', 'Survived'], survival_counts.values, color=['#e74c3c', '#2ecc71'])
plt.title('Survival Count')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.hist(df['age'], bins=30, color='#3498db', edgecolor='black')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6, 6))
class_counts = df['pclass'].value_counts()
plt.pie(class_counts.values, labels=['Class 3', 'Class 1', 'Class 2'],
        autopct='%1.1f%%', colors=['#e67e22', '#9b59b6', '#1abc9c'], startangle=140)
plt.title('Passenger Class Distribution')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='survived', data=df, palette=['#e74c3c', '#2ecc71'],legend=False)
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.title('Survival Count (Seaborn)')

plt.tight_layout()
plt.savefig('chart4_countplot_survival.png')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(x='survived', y='age', data=df, palette=['#e74c3c', '#2ecc71'])
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.title('Age Distribution by Survival')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(
    corr,annot=True,
    cmap="Reds"

)
plt.title("Heat map")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```

```python
sns.set_theme(style="darkgrid")
sns.pairplot(df,hue="sex",kind="kde",diag_kind="auto")
plt.show()
```

```python
sns.violinplot(
    data=df,
    x='embarked',
    y='fare',
    hue='embarked'
    
)
plt.title("Violin Plot")
plt.show()
```

```python
pairplot_df = df[['survived', 'age', 'fare', 'pclass', 'sibsp', 'parch']].dropna()
sns.pairplot(pairplot_df, hue='survived', palette=['#e74c3c', '#2ecc71'])
plt.suptitle('Pairplot of Key Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(x='sex', y='survived', data=df, palette=['#3498db', '#e91e8c'])
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(x='pclass', y='survived', data=df, palette='Blues_d')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Class')
plt.show()
```

```python
print("KEY INSIGHTS")
 
survival_by_gender = df.groupby('sex')['survived'].mean() * 100
survival_by_class = df.groupby('pclass')['survived'].mean() * 100
print("SURVIVAL BY PASSENGER CLASS : \n",survival_by_class)
print("SURVIVAL BY GENDER: \n",survival_by_gender)
```
