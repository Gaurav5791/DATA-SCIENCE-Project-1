Q36. Merge two DataFrames using a common key.

```python
data1_q36 = {'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']}
df1_q36 = pd.DataFrame(data1_q36)

data2_q36 = {'ID': [1, 2, 4], 'Age': [25, 30, 22]}
df2_q36 = pd.DataFrame(data2_q36)

merged_df_q36 = pd.merge(df1_q36, df2_q36, on='ID')
print(merged_df_q36)
```

Q37. Perform a left join between two DataFrames.

```python
left_merged_df_q37 = pd.merge(df1_q36, df2_q36, on='ID', how='left')
print(left_merged_df_q37)
```

Q38. Concatenate two DataFrames row-wise.

```python
data_concat1_q38 = {'Name': ['David', 'Eve'], 'Age': [28, 32]}
df_concat1_q38 = pd.DataFrame(data_concat1_q38)

data_concat2_q38 = {'Name': ['Frank', 'Grace'], 'Age': [24, 29]}
df_concat2_q38 = pd.DataFrame(data_concat2_q38)

concatenated_df_rows_q38 = pd.concat([df_concat1_q38, df_concat2_q38], ignore_index=True)
print(concatenated_df_rows_q38)
```

Q39. Concatenate two DataFrames column-wise.

```python
data_col_concat1_q39 = {'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']}
df_col_concat1_q39 = pd.DataFrame(data_col_concat1_q39)

data_col_concat2_q39 = {'Age': [25, 30, 22], 'City': ['New York', 'London', 'Paris']}
df_col_concat2_q39 = pd.DataFrame(data_col_concat2_q39)

concatenated_df_cols_q39 = pd.concat([df_col_concat1_q39, df_col_concat2_q39], axis=1)
print(concatenated_df_cols_q39)
```

Q40. Mini Project: Create a sales DataFrame and answer business questions using filtering, groupby, and aggregation.

```python
sales_data_q40 = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics'],
    'Revenue': [1200, 25, 75, 300, 1100, 20, 80, 320]
}
df_sales_q40 = pd.DataFrame(sales_data_q40)
print(df_sales_q40)
```

```python
total_revenue_per_product_q40 = df_sales_q40.groupby('Product')['Revenue'].sum()
print(total_revenue_per_product_q40)
```

```python
total_revenue_per_category_q40 = df_sales_q40.groupby('Category')['Revenue'].sum()
print(total_revenue_per_category_q40)
```

```python
high_revenue_products_q40 = df_sales_q40[df_sales_q40['Revenue'] > 100]
print(high_revenue_products_q40)
```
